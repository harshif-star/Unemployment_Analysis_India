```python
import pandas as pd

df = pd.read_csv("Unemployment in India.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    dayfirst=True,
    errors="coerce"
)

df.head()
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    dayfirst=True,
    errors="coerce"
)

print(df["Date"].head())
```

```python
print("Number of regions:", df["Region"].nunique())

print("\nRegions:")
print(df["Region"].unique())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
region_avg = (
    df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

display(region_avg)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 8))

region_avg.sort_values().plot(kind="barh")

plt.title("Average Unemployment Rate by Region")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")

plt.tight_layout()
plt.show()
```

```python
print("Observation: The average unemployment rate varies across different regions of India. The differences indicate variations in employment opportunities and labour-market conditions across regions.")
```

```python
monthly_avg = (
    df.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_avg.index,
    monthly_avg.values,
    marker="o"
)

plt.title("Unemployment Rate Trend in India")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
pre_covid = df[df["Date"] < "2020-03-01"]
post_covid = df[df["Date"] >= "2020-03-01"]

print("Pre-COVID records:", len(pre_covid))
print("Post-COVID records:", len(post_covid))
```

```python
pre_covid_avg = pre_covid[
    "Estimated Unemployment Rate (%)"
].mean()

post_covid_avg = post_covid[
    "Estimated Unemployment Rate (%)"
].mean()

print("Pre-COVID average unemployment:",
      round(pre_covid_avg, 2), "%")

print("Post-COVID average unemployment:",
      round(post_covid_avg, 2), "%")
```

```python
#Top 10 regions with highest unemployment
```

```python
top_10 = (
    df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(top_10)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10.values,
    y=top_10.index
)

plt.title("Top 10 Regions with Highest Average Unemployment Rate")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")

plt.tight_layout()
plt.show()
```

```python
#Observation: The chart shows the top 10 regions with the highest average unemployment rates during the period covered by the dataset. The variation between regions highlights differences in employment conditions across India.
```

```python
#Correlation Heatmap
```

```python
correlation_data = df[
    [
        "Estimated Unemployment Rate (%)",
        "Estimated Employed",
        "Estimated Labour Participation Rate (%)"
    ]
]

correlation_matrix = correlation_data.corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Between Employment Indicators")
plt.tight_layout()
plt.show()
```

```python
#Observation: The correlation heatmap shows the relationship between unemployment rate, estimated employment, and labour participation rate. Positive correlation indicates that two variables tend to increase together, while negative correlation indicates an inverse relationship.
```

```python
#Pre-COVID vs Post-COVID comparison
```

```python
pre_covid
post_covid
```

```python
indicators = [
    "Estimated Unemployment Rate (%)",
    "Estimated Employed",
    "Estimated Labour Participation Rate (%)"
]

pre_covid_mean = pre_covid[indicators].mean()
post_covid_mean = post_covid[indicators].mean()

comparison = pd.DataFrame({
    "Pre-COVID": pre_covid_mean,
    "Post-COVID": post_covid_mean
})

display(comparison)
```

```python
comparison.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Pre-COVID vs Post-COVID Labour Market Indicators")
plt.xlabel("Indicator")
plt.ylabel("Average Value")

plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
#Observation: The comparison shows how India's labour-market indicators changed between the pre-COVID and post-COVID periods. The unemployment rate provides an indication of the disruption caused by the pandemic, while employment and labour participation values show broader changes in labour-market activity.
```

```python
#Calculate the COVID impact specifically
```

```python
unemployment_change = (
    post_covid_mean["Estimated Unemployment Rate (%)"]
    - pre_covid_mean["Estimated Unemployment Rate (%)"]
)

print(
    "Change in average unemployment rate:",
    round(unemployment_change, 2),
    "percentage points"
)
```

```python
percentage_change = (
    unemployment_change
    / pre_covid_mean["Estimated Unemployment Rate (%)"]
) * 100

print(
    "Percentage change:",
    round(percentage_change, 2),
    "%"
)
```

```python
#This gives you an actual numerical statement for your report instead of simply saying "COVID affected unemployment"
```

```python
#Create a final summary table
```

```python
summary = pd.DataFrame({
    "Indicator": indicators,
    "Pre-COVID Average": [
        pre_covid_mean[indicator]
        for indicator in indicators
    ],
    "Post-COVID Average": [
        post_covid_mean[indicator]
        for indicator in indicators
    ]
})

display(summary)
```
